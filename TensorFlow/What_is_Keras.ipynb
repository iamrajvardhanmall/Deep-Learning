# Introduction to Keras

## Overview

Keras is a high-level deep learning API that simplifies the process of building, training, and deploying deep neural networks. It provides a user-friendly interface that allows developers and researchers to create complex deep learning models with minimal code.

Originally developed as an independent library, Keras is now fully integrated into TensorFlow as its official high-level API. This integration combines the simplicity of Keras with the scalability and performance of TensorFlow.

---

## Features of Keras

- Simple and intuitive API for building deep learning models.
- Seamlessly integrated with TensorFlow.
- Supports rapid prototyping and experimentation.
- Provides pre-built layers, activation functions, optimizers, and loss functions.
- Supports both CPU and GPU acceleration.
- Facilitates easy model training, evaluation, and deployment.
- Suitable for beginners as well as experienced deep learning practitioners.

---

## Backend Support

Historically, Keras supported multiple backend engines, including:

- TensorFlow
- Theano
- Microsoft Cognitive Toolkit (CNTK)

Today, TensorFlow is the primary and officially supported backend for Keras.

---

## Advantages of Keras

### 1. Ease of Use

Keras provides a clean and straightforward syntax, making deep learning accessible to beginners.

### 2. Faster Development

With its modular design and pre-built components, developers can quickly create and test neural network architectures.

### 3. Flexibility

Keras supports various types of neural networks, including:

- Feedforward Neural Networks (ANN)
- Convolutional Neural Networks (CNN)
- Recurrent Neural Networks (RNN)
- Long Short-Term Memory Networks (LSTM)
- Transformer-based Models

### 4. TensorFlow Integration

Since Keras is integrated into TensorFlow, users can leverage TensorFlow's powerful ecosystem, including TensorBoard, TensorFlow Lite, and TensorFlow Serving.

# How to Build a Model in Keras

Keras provides multiple ways to build deep learning models. The most commonly used approach is the **Sequential API**, which is simple, intuitive, and suitable for models with a single input and output where layers are arranged in a linear sequence.

## 1. Building a Model Using the Sequential API

The **Sequential API** allows you to create neural networks by stacking layers one after another. It is ideal for beginners and is commonly used for classification and regression tasks.

### Steps to Build a Sequential Model

1. Create a Sequential model.
2. Add a fully connected (**Dense**) layer with 64 neurons and the ReLU activation function.
3. Add an output Dense layer with 10 neurons and the Softmax activation function for multi-class classification.


### Explanation

* **Sequential()** creates an empty neural network model.
* **Dense(64, activation='relu')** adds a fully connected hidden layer with 64 neurons using the ReLU activation function.
* **input_shape=(100,)** specifies that the model expects 100 input features.
* **Dense(10, activation='softmax')** creates the output layer with 10 neurons, suitable for a classification problem with 10 classes.
* **model.summary()** displays the architecture of the neural network.

### Model Architecture

```text
Input Layer (100 Features)
          │
          ▼
Dense Layer (64 Neurons, ReLU)
          │
          ▼
Dense Layer (10 Neurons, Softmax)
          │
          ▼
Output Predictions
```

## Advantages of Sequential API

* Simple and easy to understand.
* Requires minimal code.
* Ideal for linear layer stacks.
* Suitable for beginners learning deep learning concepts.
* Fast model development and prototyping.

## Conclusion

The Sequential API is the easiest way to build neural networks in Keras. By stacking layers sequentially, developers can quickly create, train, and evaluate deep learning models for a variety of machine learning tasks.


````python
# Example
from keras.models import Sequential
from keras.layers import Dense, Activation

model = Sequential()
model.add(Dense(units=64, input_dim=100))
model.add(Activation('relu'))
model.add(Dense(units=10))
model.add(Activation('softmax'))
````

# 2. Building a Model Using the Functional API

The **Functional API** is a more flexible way to build deep learning models in Keras. Unlike the Sequential API, it allows developers to create complex neural network architectures with multiple inputs, multiple outputs, shared layers, branching structures, and skip connections.

The Functional API is particularly useful for advanced deep learning applications where a simple linear stack of layers is not sufficient.



## Explanation

* **Input()** creates input tensors for the model.
* **Dense()** adds fully connected neural network layers.
* **concatenate()** combines multiple tensors into a single tensor.
* **Model()** defines the complete neural network by specifying inputs and outputs.
* **model.summary()** displays the model architecture.

## Model Architecture

```text
Input 1 (32 Features) ──► Dense(64)
                             │
                             ▼
                         Concatenate ──► Dense(10, Softmax) ──► Output
                             ▲
                             │
Input 2 (16 Features) ──► Dense(32)
```

## Advantages of Functional API

* Supports multiple inputs and outputs.
* Enables layer sharing between different parts of a model.
* Allows creation of complex architectures such as ResNet and Inception networks.
* Supports skip connections and branching networks.
* Provides greater flexibility than the Sequential API.

## When to Use Functional API

Use the Functional API when:

* The model has multiple inputs or outputs.
* Layers need to be shared.
* The architecture is non-linear.
* Skip connections are required.
* Building advanced deep learning models.

## Conclusion

The Functional API provides a powerful and flexible approach to building neural networks in Keras. It enables the creation of sophisticated deep learning architectures that cannot be implemented using the Sequential API alone.


## Example

In this example:

1. Two input layers (`input1` and `input2`) are created.
2. Each input is processed through its own hidden layer.
3. The outputs of the hidden layers are merged using the `concatenate()` function.
4. A final output layer with Softmax activation is added for classification.



````python
from keras.layers import Input, Dense, concatenate
from keras.models import Model

input1 = Input(shape=(100,))
input2 = Input(shape=(50,))
hidden1 = Dense(64, activation='relu')(input1)
hidden2 = Dense(32, activation='relu')(input2)
merged = concatenate([hidden1, hidden2])
output = Dense(10, activation='softmax')(merged)

model = Model(inputs=[input1, input2], outputs=output)
````
