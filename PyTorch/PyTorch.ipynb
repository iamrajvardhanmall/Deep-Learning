# PyTorch

PyTorch is a Python-based deep learning library that runs on the CPU by default and supports GPU acceleration through CUDA. It follows a **define-by-run** approach, which creates dynamic computation graphs during execution, making debugging and model customization easier.

## Key Features

- Uses **dynamic computation graphs** for greater flexibility.
- Provides **automatic differentiation (Autograd)** for efficient gradient computation.
- Supports **GPU acceleration with CUDA** for faster training and inference.
- Offers a rich ecosystem for deep learning research and production applications.

## Why Use PyTorch?

PyTorch is widely used in machine learning, deep learning, NLP and Computer Vision because of its intuitive Python interface, strong community support, and seamless integration with scientific computing libraries such as NumPy and Matplotlib.

# CUDA (Compute Unified Device Architecture)

**CUDA (Compute Unified Device Architecture)** is a software platform and programming interface developed by **NVIDIA** that enables developers to use a **Graphics Processing Unit (GPU)** for general-purpose computing, not just graphics rendering.

In simple terms, CUDA acts as a bridge between software applications and NVIDIA GPUs, allowing computationally intensive tasks to be executed much faster through parallel processing.

---

## CPU vs GPU: Understanding the Difference

### CPU (Central Processing Unit)

Think of a CPU as a small team of highly skilled math professors.

- Contains a relatively small number of powerful cores (typically 4–24).
- Excels at complex decision-making and sequential tasks.
- Processes operations one after another.
- Ideal for running operating systems and general-purpose applications.

### GPU (Graphics Processing Unit)

Think of a GPU as an army of thousands of students solving simple problems simultaneously.

- Contains thousands of smaller processing cores.
- Designed for parallel processing.
- Can perform millions of simple calculations at the same time.
- Ideal for graphics rendering, machine learning, and scientific computing.

---

## Where CUDA Fits In

Originally, GPUs were designed only for graphics-related computations such as:

- Video game rendering
- 3D modeling
- Image processing

In 2006, NVIDIA introduced **CUDA**, allowing developers to use GPUs for non-graphics workloads.

With CUDA, developers can instruct the GPU to perform large-scale computations in parallel. For example:

> Instead of calculating pixels on a screen, the GPU can process millions of mathematical operations simultaneously, dramatically reducing computation time.

---

## Why CUDA is Important for Data Science and AI

Modern Artificial Intelligence and Deep Learning models require billions of mathematical operations, especially:

- Matrix multiplication
- Vector operations
- Gradient calculations
- Neural network training

A CPU may take days or weeks to complete these calculations, while a CUDA-enabled GPU can often complete them in hours or minutes.

### Benefits of CUDA in AI

- Faster model training
- Accelerated inference
- Efficient handling of large datasets
- Improved deep learning performance

---

## CUDA and Python Libraries

Data scientists rarely write CUDA code directly. Instead, popular Python libraries use CUDA behind the scenes.

Some examples include:

| Library | Purpose |
|----------|---------|
| **PyTorch** | Deep Learning Framework |
| **TensorFlow** | Machine Learning and Deep Learning |
| **CuDF** | GPU-Accelerated DataFrames (Pandas Alternative) |
| **RAPIDS** | GPU-Accelerated Data Science Toolkit |

These libraries automatically translate Python operations into CUDA instructions that can run efficiently on NVIDIA GPUs.

---

## Key Takeaways

- **CUDA** is NVIDIA's platform for GPU-accelerated computing.
- It enables GPUs to perform general-purpose computations beyond graphics rendering.
- GPUs excel at parallel processing, making them ideal for AI and machine learning workloads.
- Frameworks such as **PyTorch** and **TensorFlow** use CUDA to accelerate deep learning tasks.
- CUDA is one of the primary reasons modern AI models can be trained efficiently on large datasets.

In PyTorch, neural networks are built using the torch.nn module where:
- nn.Linear(in_features, out_features) defines a fully connected (dense) layer.
- Activation functions like torch.relu, torch.sigmoid or torch.softmax are applied between layers.
- forward() method defines how data moves through the network.

To build a neural network in PyTorch, we create a class that inherits from torch.nn.Module and defines its layers and forward pass.

````python
import torch
import torch.nn as nn

class NeuralNetwork(nn.Module):
    def __init__(self):
        super(NeuralNetwork, self).__init__()
        self.fc1 = nn.Linear(10, 16)
        self.fc2 = nn.Linear(16, 8)
        self.fc3 = nn.Linear(8, 1)

    def forward(self, x):
        x = torch.relu(self.fc1(x))
        x = torch.relu(self.fc2(x))
        x = torch.sigmoid(self.fc3(x))
        return x

model = NeuralNetwork()
print(model)
````

# Loss Function and Optimizer

After defining a neural network model, we need to specify two important components:

1. **Loss Function** – Measures how far the model's predictions are from the actual target values.
2. **Optimizer** – Updates the model's weights and biases to minimize the loss during training.

---

## Loss Function

A loss function quantifies the error between the predicted output and the true output.

For binary classification problems, we commonly use **Binary Cross-Entropy Loss (BCELoss)**:

```python
criterion = nn.BCELoss()
```

### Why BCELoss?

- Suitable for binary classification tasks.
- Compares predicted probabilities with actual labels (0 or 1).
- Produces lower loss values when predictions are closer to the true labels.

---

## Optimizer

An optimizer adjusts the model parameters based on the computed gradients obtained during backpropagation.

A widely used optimizer is **Adam (Adaptive Moment Estimation)**:

```python
optimizer = optim.Adam(model.parameters(), lr=0.001)
```

### Why Adam?

- Combines the advantages of **Momentum** and **Adaptive Learning Rates**.
- Converges faster than traditional gradient descent methods.
- Requires less hyperparameter tuning.
- Works well for most deep learning applications.

---

## Example

```python
import torch.nn as nn
import torch.optim as optim

model = NeuralNetwork()

# Loss Function
criterion = nn.BCELoss()

# Optimizer
optimizer = optim.Adam(model.parameters(), lr=0.001)
```

---

## Summary

- **Loss Function** measures prediction error.
- **BCELoss** is commonly used for binary classification problems.
- **Optimizer** updates model parameters to reduce loss.
- **Adam Optimizer** combines momentum and adaptive learning rates, making it one of the most popular optimization algorithms in deep learning.

# Training the Model

Training a neural network involves feeding data into the model, calculating the prediction error, and updating the model's weights to improve performance over time.

In this example, we generate **dummy data** consisting of:

- **100 samples**
- **10 input features** per sample
- **Binary target labels (0 or 1)**

The model is then trained using a training loop.

````python
import torch.optim as optim

# 1. Define the Loss Function
criterion = nn.BCELoss() 

# 2. Define the Optimizer (linking it to your model's parameters)
optimizer = optim.Adam(model.parameters(), lr=0.01) 

inputs = torch.randn((100, 10))
targets = torch.randint(0, 2, (100, 1)).float()
epochs = 20

for epoch in range(epochs):
    optimizer.zero_grad()
    outputs = model(inputs)
    loss = criterion(outputs, targets)
    loss.backward()
    optimizer.step()

    if (epoch + 1) % 5 == 0:
        print(f"Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}")
````

---

## Training Process

During each training epoch, the following steps are performed:

### 1. Clear Previous Gradients

```python
optimizer.zero_grad()
```

- Clears the gradients accumulated from the previous iteration.
- Prevents gradients from being added repeatedly during training.

---

### 2. Forward Pass

```python
outputs = model(inputs)
```

- Passes the input data through the neural network.
- Generates predictions based on the current model weights.

---

### 3. Compute Loss

```python
loss = criterion(outputs, targets)
```

- Measures the difference between predicted outputs and actual target values.
- A lower loss indicates better model performance.

---

### 4. Backpropagation

```python
loss.backward()
```

- Computes gradients of the loss with respect to all trainable parameters.
- Uses the chain rule to determine how each weight contributed to the error.

---

### 5. Update Model Parameters

```python
optimizer.step()
```

- Updates the model weights and biases using the computed gradients.
- Moves the parameters in the direction that minimizes the loss.

---

## Example Training Loop

```python
for epoch in range(num_epochs):

    # Clear previous gradients
    optimizer.zero_grad()

    # Forward pass
    outputs = model(inputs)

    # Compute loss
    loss = criterion(outputs, targets)

    # Backpropagation
    loss.backward()

    # Update weights
    optimizer.step()

    if (epoch + 1) % 5 == 0:
        print(f"Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}")
```

---

## Summary

- `optimizer.zero_grad()` removes previously stored gradients.
- `model(inputs)` performs the forward pass and generates predictions.
- `criterion(outputs, targets)` computes the prediction error.
- `loss.backward()` calculates gradients using backpropagation.
- `optimizer.step()` updates model parameters to reduce the loss.
- Repeating these steps over multiple epochs enables the neural network to learn patterns from the data.
